## 2. Núcleo Teórico: Mecanismos de Crecimiento (40 min)



* **Crecimiento por Niveles (Level-wise / Depth-wise):**
* Estrategia de recorrido a lo ancho (*Breadth-First*).
* Divide todos los nodos de un nivel simultáneamente, manteniendo el árbol balanceado.
* Desventaja: Gasta recursos computacionales dividiendo nodos con bajo gradiente (poca ganancia de información), lo cual suaviza demasiado los picos de las series de tiempo.




* **Crecimiento por Hojas (Leaf-wise):**
* Estrategia de recorrido en profundidad (*Depth-First*).
* Busca de forma codiciosa (*greedy*) la hoja que ofrece la mayor reducción de la función de pérdida (mayor ganancia), sin importar el nivel en el que se encuentre.
* Ventaja: Logra converger de forma mucho más rápida y modela mejor las asimetrías y cambios abruptos de las epidemias.





### 3. La Magia de LightGBM: Optimización por Histogramas (30 min)

* Sustitución del escaneo continuo de valores flotantes (temperatura, humedad, lluvias) por **contenedores discretos (Bins)**.
* Reducción del costo de encontrar el punto de división (*split point*) de $\mathcal{O}(\#\text{datos} \times \#\text{atributos})$ a $\mathcal{O}(\#\text{bins} \times \#\text{atributos})$.
* Cómo esta discretización actúa como un regularizador natural que evita que el ruido meteorológico diario confunda al modelo de predicción semanal.



# 4. Debate y Aplicación Práctica al Nexo Clima-Salud (20 min)

* **¿Por qué Leaf-wise es óptimo para datos exógenos?**
* Las variables climáticas tienen interacciones no lineales complejas (ej. *Alta Temperatura* **Y** *Alta Humedad* prolongada desatan la eclosión del vector). El crecimiento *Leaf-wise* aísla estas interacciones críticas rápidamente en ramas profundas y específicas, mientras ignora combinaciones irrelevantes.




* **El Peligro Inherente:** Mayor tendencia al sobreajuste (*overfitting*) en muestras pequeñas si no se controlan `max_depth` y `num_leaves`.

## Cronograma de la Sesión (120 minutos)

### 1. Introducción y el Problema de la Dimensionalidad Climática (20 min)

* El desafío de los modelos basados en árboles cuando multiplicamos variables exógenas (ej. 10 variables meteorológicas $\times$ 12 rezagos temporales = 120 predictores correlacionados).
* ¿Por qué los árboles de decisión tradicionales (RF, Gradient Boosting clásico) se vuelven lentos e ineficientes ante estas matrices?


Para abrir la **Clase 1.1** con un impacto visual y práctico inmediato, este ejemplo en Python simula el escenario exacto del problema: una matriz densa de rezagos meteorológicos altamente correlacionados (multicolinealidad) que satura a los algoritmos tradicionales.

El script genera datos sintéticos que imitan la realidad epidemiológica, entrena un **Random Forest clásico** vs. **LightGBM**, y mide el tiempo de ejecución. Esto le demostrará a tus estudiantes, con datos en mano, el cuello de botella computacional antes de entrar a la teoría.


```python
import time
import numpy as np
import pandas as pd
import lightgbm as lgb
from sklearn.ensemble import RandomForestRegressor

# =============================================================================
# SIMULACIÓN DEL PROBLEMA: ALTA DIMENSIONALIDAD Y MULTICOLINEALIDAD CLIMÁTICA
# =============================================================================
# =============================================================================
# CORRECCIÓN PARA LA CLASE: ESCALA REAL DE DATOS (Big Data / Datos Diarios)
# =============================================================================
print("=== CONFIGURANDO EL ESCENARIO CORREGIDO DE LA CLASE 1.1 ===")
np.random.seed(42)

# Simulamos datos diarios o múltiples estaciones climatológicas (10,000 registros)
n_registros = 10000 

n_vars_meteorologicas = 10
# Usamos una caminata aleatoria para que SÍ exista correlación temporal real entre lags
meteo_base = np.zeros((n_registros, n_vars_meteorologicas))
for i in range(1, n_registros):
    meteo_base[i] = meteo_base[i-1] * 0.9 + np.random.normal(2, 1, n_vars_meteorologicas)

y = np.log1p(np.maximum(0, meteo_base[:, 0] * 1.5 + np.random.normal(0, 5, n_registros)))

# Creamos los mismos 120 predictores con rezagos climáticos
X_list = []
for lag in range(1, 13):
    df_lag = pd.DataFrame(meteo_base).shift(lag)
    df_lag.columns = [f'meteo_var_{i}_lag_{lag}' for i in range(n_vars_meteorologicas)]
    X_list.append(df_lag)

X = pd.concat(X_list, axis=1).dropna()
y = y[X.index]

print(f"[CORREGIDO] Tamaño de X: {X.shape[0]} registros x {X.shape[1]} variables.")
print(f"[CORREGIDO] Correlación real simulada (Lag 1 vs Lag 2): {X.iloc[:,0].corr(X.iloc[:,10]):.4f}\n")

# ... (Ejecuta aquí el mismo entrenamiento de RF y LightGBM del script anterior)


# =============================================================================
# PRUEBA DE RENDIMIENTO: ÁRBOL CLÁSICO VS LIGHTGBM
# =============================================================================

# --- 1. Algoritmo Tradicional: Random Forest Regressor ---
# Los árboles tradicionales evalúan de forma continua todos los posibles puntos de corte en las 120 variables
print("Entrenando Random Forest Tradicional (Scikit-Learn)...")
rf_modelo = RandomForestRegressor(n_estimators=100, max_depth=5, random_state=42, n_jobs=1)

inicio_rf = time.time()
rf_modelo.fit(X, y)
tiempo_rf = time.time() - inicio_rf
print(f"[RESULTADO] Random Forest tardó: {tiempo_rf:.4f} segundos.")

# --- 2. Algoritmo Optimizado: LightGBM ---
# Gracias al crecimiento Leaf-wise y el mapeo por Histogramas, ignora el ruido de la multicolinealidad
print("\nEntrenando LightGBM Regressor (Optimizado por Histogramas)...")
lgb_modelo = lgb.LGBMRegressor(n_estimators=100, max_depth=5, num_leaves=31, random_state=42, verbosity=-1)

inicio_lgb = time.time()
lgb_modelo.fit(X, y)
tiempo_lgb = time.time() - inicio_lgb
print(f"[RESULTADO] LightGBM tardó: {tiempo_lgb:.4f} segundos.")

# --- 3. Comparativa Final ---
print("\n" + "="*50)
print(f" LightGBM fue {tiempo_rf / tiempo_lgb:.1f}x más rápido que Random Forest")
print("="*50)
```

Acabas de toparde con un escenario real y clásico en la práctica de la ciencia de datos. Obtuviste un resultado inverso (donde **Random Forest fue 3 veces más rápido que LightGBM**) debido a dos factores críticos que ocurren al trabajar con conjuntos de datos muy pequeños: la **escala de los datos** y el **costo de inicialización (*overhead*)** del algoritmo.



A continuación, te explico técnicamente por qué falló la simulación en este caso y cómo corregirlo para que tus alumnos vean la verdadera ventaja de LightGBM.




## 1. La razón del resultado: Datos Demasiado Pequeños

Los algoritmos avanzados como LightGBM están diseñados como "fórmula 1" para el procesamiento de datos: vuelan en autopistas largas, pero sufren en estacionamientos pequeños.

### El fenómeno del *Overhead* (Costo de Inicialización)

LightGBM es una librería externa muy compleja escrita en C++. Cada vez que ejecutas `.fit()`, el algoritmo debe realizar internamente tareas pesadas de preparación:

1. Construir las estructuras para los **histogramas**.
2. Configurar los hilos de procesamiento en paralelo.
3. Reservar memoria caché optimizada.

En una matriz de apenas **144 filas** (semanas), ese proceso de inicialización tarda mucho más tiempo que el entrenamiento en sí. En cambio, el `Random Forest` de Scikit-Learn está sumamente optimizado en Cython para arreglos pequeños de NumPy; al no tener que construir histogramas ni subestructuras complejas, ejecuta sus árboles planos de forma casi instantánea en matrices diminutas.

### La falta de correlación en los datos sintéticos

Tu resultado muestra una correlación de apenas **0.0526** entre el Lag 1 y el Lag 2. En la vida real, el clima es altamente persistente (si hoy hace 30°C, mañana no hará 0°C), por lo que la correlación real suele superar el **0.85**.
Al ser datos puramente aleatorios (ruido blanco), Random Forest encuentra divisiones de forma caótica y rápida, mientras que LightGBM gasta tiempo buscando patrones en histogramas que están vacíos o planos.
